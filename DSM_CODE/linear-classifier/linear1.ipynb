```python
import pandas as pd
from sklearn.linear_model import Perceptron
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({'X':[-1,1,2], 'Y':[-1,1,1]})
df
```

```python
plt.scatter(df['X'], df['Y'])
plt.show()
```

```python
import numpy as np
model = Perceptron(fit_intercept=False, eta0=0.01)
model.coef_ = np.array([[-5]])
model = model.fit(df[['X']], df['Y'])
```

```python
model.coef_
```

```python
import numpy as np
plt.scatter(df['X'], df['Y'])
# disenga una linea reatta con coefficiente model.coef_[0][0]
x_vals = np.array(plt.gca().get_xlim())
y_vals = model.coef_[0][0] * x_vals
plt.plot(x_vals, y_vals, color='red')
# disegna la linea y = 0
plt.axhline(0, color='black', lw=1, ls='--')
# disegna la linea x = 0
plt.axvline(0, color='black', lw=1, ls='--')
plt.show()
```

```python
# calcola accuratezza
from sklearn.metrics import accuracy_score
y_pred = model.predict(df[['X']])
accuracy = accuracy_score(df['Y'], y_pred)  
accuracy
```
